# Analysis

```python
event_id = '5-6'
electrode_group = p3a
```

```python
epochs_dis = mne.read_epochs (path_to_ALL_ALL_base_200_onset_epoches/'_epo.fif', preload = True)
epochs_dis = epochs_dis [event_id]
# Get indices of the selected channels
ch_idxs = [epochs_dis.info['ch_names'].index(ch) for ch in electrode_group]
X = epochs_dis.get_data()  # as 3D matrix
X = X[:, ch_idxs, :]  # Subset data for selected channels: (n_epochs, n_selected_channels, n_times)
X = np.mean(X, axis=1)  # Average across channels: (n_epochs, n_times)
```

```python
T, pval = stats.ttest_1samp(X, 0)
alpha = 0.05

n_samples, n_tests = X.shape
threshold_uncorrected = stats.t.ppf(1.0 - alpha, n_samples - 1)

reject_bonferroni, pval_bonferroni = bonferroni_correction(pval, alpha=alpha)
threshold_bonferroni = stats.t.ppf(1.0 - alpha / n_tests, n_samples - 1)

reject_fdr, pval_fdr = fdr_correction(pval, alpha=alpha, method="indep")
# Check if any points survive FDR correction
if np.any(reject_fdr):
    threshold_fdr = np.min(np.abs(T)[reject_fdr])
else:
    threshold_fdr = None  # or set to np.inf if you prefer
```

```python
times = 1e3 * epochs_dis.times

plt.close("all")
plt.plot(times, T, "k", label="T-stat")
xmin, xmax = plt.xlim()
plt.hlines(
    threshold_uncorrected,
    xmin,
    xmax,
    linestyle="--",
    colors="k",
    label="p=0.05 (uncorrected)",
    linewidth=2,
)
plt.hlines(
    threshold_bonferroni,
    xmin,
    xmax,
    linestyle="--",
    colors="r",
    label="p=0.05 (Bonferroni)",
    linewidth=2,
)
plt.hlines(
    threshold_fdr,
    xmin,
    xmax,
    linestyle="--",
    colors="b",
    label="p=0.05 (FDR)",
    linewidth=2,
)
plt.legend()
plt.xlabel("Time (ms)")
plt.ylabel("T-stat")
plt.show()
plt.title(f"t-test with multiple correction {event_id} {electrode_group}.png")
# Save the plot with a high resolution
filename = f"t-test with multiple correction {event_id} {electrode_group}.png"
plt.savefig(path_to_correction_graphs / filename, dpi=300)  # Save with high resolution
```
