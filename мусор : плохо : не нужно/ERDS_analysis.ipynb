# Analysis one graph for each channel in the list

```python
freqs = np.arange(1, 81)  # frequencies from 2-35Hz
vmin, vmax = -0.5, 2.0  # set min and max ERDS values in plot
baseline = (-0.5, 0)  # baseline interval (in s)
cnorm = TwoSlopeNorm(vmin=vmin, vcenter=0, vmax=vmax)  # min, center & max ERDS
tmin, tmax = -0.5, 2
```

```python
kwargs = dict(
    n_permutations=100, step_down_p=0.05, seed=1, buffer_size=None, out_type="mask"
)  # for cluster test
```

```python
tfr = epochs_dis.compute_tfr(
    method="multitaper",
    freqs=freqs,
    n_cycles=freqs,
    use_fft=True,
    return_itc=False,
    average=False,
    decim=2,
    picks=temporalL
)
tfr.crop(tmin, tmax).apply_baseline(baseline, mode="percent")
```

```python
# Assuming dis_ids is defined and contains the relevant event IDs
for event in dis_ids:
    # Select the grand average TFR for the current event
    tfr_ev = tfr[event]
    
    for ch in temporalL:  # Loop through each channel in frontalL

        # Find the index of the channel name
        ch_idx = tfr_ev.info['ch_names'].index(ch)  # Use the averaged channel

        fig, ax = plt.subplots(figsize=(6, 4))  # Create a new figure for each channel
        
        # Positive clusters
        _, c1, p1, _ = pcluster_test(tfr_ev.data[:, ch_idx], tail=1, **kwargs)
        # Negative clusters
        _, c2, p2, _ = pcluster_test(tfr_ev.data[:, ch_idx], tail=-1, **kwargs)

        # Combine clusters from the positive and negative tests
        c = np.stack(c1 + c2, axis=2)  # Combined clusters
        p = np.concatenate((p1, p2))  # Combined p-values
        mask = c[..., p <= 0.05].any(axis=-1)  # Create mask for significant clusters

        # Plot TFR (ERDS map with masking)
        tfr_ev.average().plot(
            picks=[ch_idx],
            cmap="RdBu",
            cnorm=cnorm,
            axes=ax,
            colorbar=True,  # Include colorbar in each plot
            show=False,
            mask=mask,
            mask_style="mask",
        )

        ax.set_title(f"{ch} (Event: {event})", fontsize=10)
        ax.axvline(0, linewidth=1, color="black", linestyle=":")  # Event line

        plt.show()  # Display each figure separately

        # Save the plot with a high resolution
        filename = f"{ch} (Event: {event}).png"
        plt.savefig(path_to_ERDS / filename, dpi=300)  # Save with high resolution
```

```python
freqs = np.arange(1, 81)  # frequencies from 2-35Hz
vmin, vmax = 0.3, 2.5  # set min and max ERDS values in plot
baseline = (0.3, 0.5)  # baseline interval (in s)
cnorm = TwoSlopeNorm(vmin=vmin, vcenter=0.5, vmax=vmax)  # min, center & max ERDS
tmin, tmax = 0.3, 2.5
```

```python
kwargs = dict(
    n_permutations=100, step_down_p=0.05, seed=1, buffer_size=None, out_type="mask"
)  # for cluster test
```

```python
# Step 1: Use the picks parameter to select the specific channels by their names
picks = mne.pick_channels(epochs_dis.info['ch_names'], include=frontalL)

# Step 2: Extract the data for the selected channels
# The shape of the data will be (n_epochs, n_selected_channels, n_times)
data_selected_channels = epochs_dis.get_data(picks=picks)

# Step 3: Average across the selected channels (axis 1)
# The resulting shape will be (n_epochs, 1, n_times)
data_avg_selected_channels = np.mean(data_selected_channels, axis=1, keepdims=True)

# Step 4: Create a new `info` structure for the averaged channel
info = mne.create_info(
    ch_names=['1'],  # Name for the averaged channel
    sfreq=epochs_dis.info['sfreq'],  # Use the same sampling frequency
    ch_types='eeg'  # Keep the same channel type
)

# Step 5: Create a new Epochs object using the averaged data
epochs_avg = mne.EpochsArray(data_avg_selected_channels, info, events=epochs_dis.events, event_id=epochs_dis.event_id)
```

```python
tfr = epochs_avg.compute_tfr(
    method="multitaper",
    freqs=freqs,
    n_cycles=freqs,
    use_fft=True,
    return_itc=False,
    average=False,
    decim=2
    )
# ОН ТЕПЕРЬ ЗА НОЛЬ СЧИТАЕТ ИНТЕРВАЛ СТАРТОВУЮ ТОЧКУ ЭПОХИ, ТО ЕСТЬ -0.5 ИЛИ -0.2 ЭТО ТЕПЕРЬ НОЛЬ
tfr.crop(tmin, tmax).apply_baseline(baseline, mode="percent")
```
